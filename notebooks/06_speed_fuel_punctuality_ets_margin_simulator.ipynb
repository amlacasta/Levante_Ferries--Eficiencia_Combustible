# 06 · Speed–fuel–punctuality–ETS–margin simulator

Explora velocidades factibles para un viaje manteniendo una restricción de puntualidad OTR15. El coste ETS es un accrual analítico, no un cálculo legal de cumplimiento.

```python
from pathlib import Path
import sys, json, numpy as np, pandas as pd
ROOT=next((p for p in [Path.cwd(), *Path.cwd().parents] if (p/'src/project8.py').is_file()), None)
if ROOT is None:
    raise FileNotFoundError('Abre el notebook dentro del repositorio del Proyecto 8.')
sys.path.insert(0,str(ROOT))
from src.project8 import *

df=pd.read_csv(ROOT/'data/raw/voyages_2026_synthetic.csv',parse_dates=['date'])
row=df.iloc[(df.load_factor-df.load_factor.median()).abs().argsort().iloc[0]]
grid,best=optimizer_grid(row,eua_price=80,fuel_price=700)
display(grid.head())
print('Voyage:',row.voyage_id,row.route_name,row.vessel_id)
print('Best feasible speed:',round(best.speed_kn,2),'kn | fuel',round(best.fuel_tonnes,2),'t | delay',round(best.delay_min,1),'min | margin €',round(best.margin_eur,0))
```

La curva refleja la relación no lineal velocidad–potencia: bajar velocidad reduce consumo, pero puede romper puntualidad. El óptimo depende del buffer de horario, mar, carga, precio del combustible y EUA.
