# 04 · SHAP explainability

Explica el modelo de consumo. SHAP aporta contribuciones locales y globales; no convierte correlaciones en causalidad.

```python
from pathlib import Path
import sys, json, numpy as np, pandas as pd
ROOT=next((p for p in [Path.cwd(), *Path.cwd().parents] if (p/'src/project8.py').is_file()), None)
if ROOT is None:
    raise FileNotFoundError('Abre el notebook dentro del repositorio del Proyecto 8.')
sys.path.insert(0,str(ROOT))
from src.project8 import *

import shap, matplotlib.pyplot as plt
df=pd.read_csv(ROOT/'data/raw/voyages_2026_synthetic.csv',parse_dates=['date'])
model,X,metrics,preds=fit_xgb(df)
Xs=X.iloc[np.linspace(0,len(X)-1,500,dtype=int)]
ex=shap.TreeExplainer(model); sv=ex.shap_values(Xs)
imp=pd.DataFrame({'feature':X.columns,'mean_abs_shap':np.abs(sv).mean(axis=0)}).sort_values('mean_abs_shap',ascending=False)
display(imp.head(15))
```

La explicación es una herramienta de control del modelo: permite comprobar si dominan factores físicamente plausibles como velocidad, distancia, meteorología y carga.
